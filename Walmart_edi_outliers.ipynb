```python
df["Weekly_Sales"].describe()
```

```python
Q1=df["Weekly_Sales"].quantile(0.25)
Q3=df["Weekly_Sales"].quantile(0.75)
Q1,Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
lower_limit,upper_limit
```

```python
df_cleaned = df[(df["Weekly_Sales"] >= lower_limit) & (df["Weekly_Sales"] <= upper_limit)]
df_cleaned
```

```python
df["Temperature"].describe()
```

```python
Q1=df["Temperature"].quantile(0.25)
Q3=df["Temperature"].quantile(0.75)
Q1,Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
lower_limit,upper_limit
```

```python
df_cleaned = df[(df["Temperature"] >= lower_limit) & (df["Temperature"] <= upper_limit)]
df_cleaned
```

```python
from sklearn.preprocessing import StandardScaler


numerical_cols = ['Weekly_Sales', 'Temperature', 'CPI', 'Unemployment']

scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_cleaned[numerical_cols])
df_scaled
```

```python
df
```

```python
df["CPI"]=df["CPI"].clip(lower=lower_limit,upper=upper_limit)
df
```

```python
df["Unemployment"]=df["Unemployment"].clip(lower=lower_limit,upper=upper_limit)
df
```
